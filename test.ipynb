```python
from fuzzywuzzy import fuzz

def find_best_match(string1, string2):
    ratio = fuzz.ratio(string1, string2)
    return ratio

def find_most_similar_string(strings):
    max_ratio = 0
    best_match = None

    for i in range(len(strings)):
        for j in range(i + 1, len(strings)):
            ratio = find_best_match(strings[i], strings[j])
            if ratio > max_ratio:
                max_ratio = ratio
                best_match = (strings[i], strings[j])

    return best_match

# Example usage
input_strings = [
    "Hello world",
    "Hello",
    "World",
    "Hi there",
    "Hola mundo"
]

best_match = find_most_similar_string(input_strings)
print("Best match:", best_match)
```
